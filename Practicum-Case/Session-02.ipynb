# Import Library

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag, ne_chunk
from nltk.probability import FreqDist
```

# Download NLTK Resources

```python
def download_nltk_resources():
    resources = [
        'punkt', 'punkt_tab', 'stopwords', 'wordnet', 
        'averaged_perceptron_tagger', 'averaged_perceptron_tagger_eng',
        'maxent_ne_chunker', 'maxent_ne_chunker_tab', 'words'
    ]
    for res in resources:
        try:
            nltk.data.find(f'tokenizers/{res}')
        except LookupError:
            try:
                nltk.data.find(f'corpora/{res}')
            except LookupError:
                try:
                    nltk.data.find(f'help/{res}')
                except LookupError:
                     try:
                        nltk.data.find(f'chunkers/{res}')
                     except LookupError:
                        print(f"Mengunduh {res}...")
                        nltk.download(res, quiet=True)

download_nltk_resources()
```

# Main Function

```python
def main():
    print("--- NLTK PROCESSOR ---")
    
    # 1. Input Validation 
    sentence = ""
    while not sentence.strip():
        sentence = input("Input sentence(s): ")
        if not sentence.strip():
            print("Input tidak boleh kosong. Silakan coba lagi.")
    
    print("\n" + "="*30 + "\n")

    # 2. Sentence Tokenization 
    # Memecah teks menjadi daftar kalimat.
    print("Your tokenized sentences are:\n")
    tokenized_sentences = sent_tokenize(sentence)
    for s in tokenized_sentences:
        print(s)
    
    print("\nnext -> tokenized words\n")

    # 3. Word Tokenization 
    # Memecah kalimat menjadi daftar kata (token).
    print("Your tokenized words are:\n")
    tokenized_words = word_tokenize(sentence)

    # Format output menggunakan tanda panah ' -> ' sesuai gambar soal
    print(" -> ".join(tokenized_words))
    
    print("\nnext -> stop words\n")

    # 4. Stop Words 
    # Mengidentifikasi kata-kata umum yang sering diabaikan (stopwords).
    print("Your stop words are:\n")
    stop_words = set(stopwords.words('english'))
    found_stopwords = [w for w in tokenized_words if w.lower() in stop_words]
    
    # Menampilkan stop words secara vertikal sesuai gambar soal
    for sw in found_stopwords:
        print(sw)
        
    print("\nnext -> stemmed words\n")

    # 5. Stemming 
    # Mengubah kata ke bentuk dasarnya (root) dengan memotong imbuhan.
    print("Your stemmed words are:\n")
    ps = PorterStemmer()
    stemmed_words = []
    
    # Format: Kata Asli -> Kata Stemmed
    for w in tokenized_words:
        root = ps.stem(w)
        stemmed_words.append(root)
        print(f"{w} -> {root}")
        
    print("\nnext -> lemmatized words\n")

    # 6. Lemmatizing 
    # Mengubah kata ke bentuk kamus (lemma) yang valid.
    print("Your lemmatized words are:\n")
    lemmatizer = WordNetLemmatizer()
    lemmatized_words = []
    
    # Format: Kata Asli -> Kata Lemma
    for w in tokenized_words:
        lemma = lemmatizer.lemmatize(w)
        lemmatized_words.append(lemma)
        print(f"{w} -> {lemma}")

    print("\nnext -> Part-of-Speech words\n")

    # 7. Part-of-Speech (POS) Tagging 
    # Memberikan label gramatikal (misal: Noun, Verb) pada setiap kata.
    print("Your Part-of-Speech words are:\n")
    pos_tags = pos_tag(tokenized_words)
    
    # Format: Kata : TAG
    for word, tag in pos_tags:
        print(f"{word} : {tag}")

    print("\nnext -> Named Entity Recognitions\n")

    # 8. Named Entity Recognition (NER) 
    # Mengidentifikasi entitas bernama dan struktur kalimat.
    # Note: Gambar soal menunjukkan visualisasi tree. 
    # Di console, kita print struktur tree-nya, dan mencoba membuka visual window jika memungkinkan.
    print("Membuka window visualisasi tree (jika environment mendukung)...")
    ner_tree = ne_chunk(pos_tags)
    print(ner_tree) # Menampilkan representasi text dari tree
    
    try:
        ner_tree.draw() 
    except Exception as e:
        print("(Visualisasi grafis tidak dapat ditampilkan di environment ini)")

    print("\nnext -> Frequency Distributions\n")

    # 9. Frequency Distributions 
    # Menghitung kemunculan setiap kata.
    print("Your Frequency Distributions are:\n")
    fdist = FreqDist(tokenized_words)
    
    # Menampilkan hasil hitungan
    for word, frequency in fdist.most_common():
        print(f"{word} = {frequency}")

    print("\n" + "="*30 + "\n")
    
    # 10. Exit [cite: 24]
    print("Thank you")

if __name__ == "__main__":
    main()
```
