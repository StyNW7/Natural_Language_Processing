```python
import nltk
import pandas as pd
import string
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords, wordnet
from nltk.stem import WordNetLemmatizer, SnowballStemmer
from nltk.tag import pos_tag
from nltk.probability import FreqDist
from nltk.classify import NaiveBayesClassifier, accuracy
import pickle

nltk.download('stopwords')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
```

```python
reviews = []

def extract_features(x):
  features = {}

  for word in freq_dist.keys():
    features[word] = word in x

  return features

stemmer = SnowballStemmer('english')
def preprocess_sentence(sentence):
  eng_stopwords = stopwords.words()
  punctuations = string.punctuation

  tokens = []
  for word in word_tokenize(sentence):
    word = word.lower()
    if word not in eng_stopwords and word not in punctuations and word.isalpha():
      tokens.append(word)


  # Bisa ditambahin stemming
  stemmed = []
  for word in tokens:
    stemmed.append(stemmer.stem(word))

  tagged = pos_tag(stemmed)
  # tagged = pos_tag(tokens)
  lemmatized = lemmatizing(tagged)

  return lemmatized

lemmatizer = WordNetLemmatizer()
def lemmatizing(tagged):
  # JJ NN VB RB
  # a n v r

  lemmatized = []
  for word, tag in tagged:
    tag = get_tag(tag)
    lemmatized.append(lemmatizer.lemmatize(word, tag))

  return lemmatized

def get_tag(tag):
  if tag == 'JJ':
    return 'a'

  if tag in ['VB', 'NN', 'RB']:
    return tag[0].lower()

  return 'n'

data = pd.read_csv('Tweets.csv')

x = data['text']
y = data['airline_sentiment']

all_data = ' '.join(x)
preprocessed_data = preprocess_sentence(all_data)
freq_dist = FreqDist(preprocessed_data)

data_pairs = zip(x,y)

feature_sets = []
for x,y in data_pairs:
  features = extract_features(preprocess_sentence(x))
  feature_sets.append((features, y))

training_count = int(len(feature_sets) * 0.8)
training_data = feature_sets[:training_count]
test_data = feature_sets[training_count:]

```

```python
#  Classifiers
try:
  with open('model.pickle', 'rb') as file:
          classifier = pickle.load(file)
          print("Model loaded!")
except:
  # makeModel()
  classifier = NaiveBayesClassifier.train(training_data)
  test_accuracy = accuracy(classifier, test_data)
  print(test_accuracy)

  with open('model.pickle', 'wb') as file:
    pickle.dump(classifier, file)
  print('Model made!')

classifier.show_most_informative_features(5)

def writeTweet():
  while True:
        tweet = input(">> ")
        count = tweet.split()

        if(len(count) > 4):
            reviews.append(tweet)
            print(tweet)
            print(reviews)
            return

        else:
            print("Invalid input. Tweet must consists of atleast 5 words.")
            print("Please try again.")
            print("")
            return

def analyze_tweet():
  if len(reviews) <= 0:
    print("No tweets found.")
    return

  for i in reviews:
    print(f'Text: {i}')

    tokens = word_tokenize(i)

    # POS Tagging
    tagged = pos_tag(tokens)
    print('Tweet POS Tags:')
    for word, tag in tagged:
      print(f'{word} : {tag}')

    # Antonyms & Synonyms
    for word in tokens:
        print(f'Word: {word}')

        synsets = wordnet.synsets(word)
        synonyms = []
        antonyms = []

        for synset in synsets:
            for synonym in synset.lemmas():
                synonyms.append(synonym.name())

                for antonym in synonym.antonyms():
                    antonyms.append(antonym.name())

        print("Synonyms:")
        if (len(synonyms) == 0):
            print("No synonyms found.")
        else:
            for word in synonyms[:5]:
                print(f"(+) {word}")

        print("")
        print("Antonyms:")
        if (len(antonyms) == 0):
            print("No antonyms found.")
        else:
            for word in antonyms[:5]:
                print(f"(-) {word}")

    print(f'Prediction: {classifier.classify(extract_features(preprocess_sentence(i)))}')

```

```python
def mainMenu():
  while True:
    print("")
    print("Menu")
    print("1. Write Tweet")
    print("2. Analyze Tweet")
    print("3. Exit")
    print("Input Your Choice:")

    choice = int(input(">> "))

    print("")

    if (choice == 1):
        print("Type your Tweet:")
        writeTweet()

        print("Tweet successfully saved!")

    elif (choice == 2):
        print("Analyzing. . .")
        print("")
        analyze_tweet()

    elif (choice == 3):
        print("Terminating Application. . .")
        break

    else:
        print("Invalid choice. Please try again.")
        print("")

mainMenu()
```
