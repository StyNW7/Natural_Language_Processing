# Stemming

## Porter Stemmer

```python
import nltk
from nltk.stem import PorterStemmer

ps = PorterStemmer()

words = ["running", "studies", "better", "playing"]

for word in words:
    print(word, "->", ps.stem(word))
```

## Lancaster Stemmer

```python
from nltk.stem import LancasterStemmer

ls = LancasterStemmer()

for word in words:
    print(word, "->", ls.stem(word))
```

## Snowball Stemmer

```python
from nltk.stem import SnowballStemmer

ss = SnowballStemmer("english")

for word in words:
    print(word, "->", ss.stem(word))
```

## WordNet Lemmatizer

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["running", "studies", "better", "wolves"]

for word in words:
    print(word, "->", lemmatizer.lemmatize(word))
```

## Lemmatizing with POS

```python
from nltk.corpus import wordnet

lemmatizer.lemmatize("running", pos=wordnet.VERB)
```

## Mapping POS From NLTK

```python
nltk.download('averaged_perceptron_tagger')

from nltk import pos_tag, word_tokenize

sentence = "The wolves are running better than dogs"
tokens = word_tokenize(sentence)

print(pos_tag(tokens))
```

# Wordnet

## Synonym

```python
from nltk.corpus import wordnet as wn

synonyms = []

for syn in wn.synsets("happy"):
    for lemma in syn.lemmas():
        synonyms.append(lemma.name())

print(set(synonyms))
```

## Antonym

```python
antonyms = []

for syn in wn.synsets("happy"):
    for lemma in syn.lemmas():
        if lemma.antonyms():
            antonyms.append(lemma.antonyms()[0].name())

print(set(antonyms))
```
